# Letters Game
*A fun word guessing game!*

```python

import random
import string

# Variables to adjust the gameplay
words_file_path = 'safedict_full.txt'
min_word_len = 15

letter_points_dict = {
    'A':1, 'E':1, 'I':1, 'O':1, 'U':1, 'L':1, 'N':1, 'S':1,
    'T':1, 'R':1, 'D':2, 'G':2, 'B':3, 'C':3, 'M':3, 'P':3,
    'F':4, 'H':4, 'V':4, 'W':4, 'Y':4, 'K':5, 'J':8, 'X':8,
    'Q':10, 'Z':10
}

def load_words():
    word_list = []
    # TODO 2: Read words from the file
    with open(words_file_path, 'r') as file:
        for line in file:
            word = line.strip().upper()
            if len(word) >= min_word_len and word.isalpha():
                word_list.append(word)
    return word_list


def print_welcome():
    # TODO 1: Print welcome message
    print("Welcome to the Word Guessing Game!")


def print_instructions():
    # TODO 8: Print instructions
    print("Guess the letters in the mystery word.")
    print("Correct guesses increase your score.")
    print("Wrong guesses decrease your score.")
    print("Type one letter at a time.")
    print("Type 'quit' to exit the game.")


def display_word(word, guess_list):
    # TODO 5: Display guessed letters and underscores
    for letter in word:
        if letter in guess_list:
            print(letter, end=" ")
        else:
            print("_", end=" ")
    print()


def is_letter(x):
    return len(x) == 1 and x in string.ascii_letters


def is_game_over(word, guess_list):
    for letter in word:
        if letter not in guess_list:
            return False
    return True


def main():
    print_welcome()
    print_instructions()

    # TODO 3: Load words
    word_list = load_words()

    # TODO 4: Choose a random word
    word = random.choice(word_list)

    guess_list = []
    score = 0

    while True:
        # TODO 6: Display the word
        display_word(word, guess_list)

        print('You have guessed: ' + str(guess_list))
        print('Your score is ' + str(score))

        # TODO 7: Get user input
        user_input = input("Enter a letter or type 'quit': ").lower()

        if user_input == 'quit':
            print("You quit the game.")
            break

        elif is_letter(user_input):
            user_input = user_input.upper()

            if user_input in guess_list:
                print('You have already guessed this letter.')
            else:
                if user_input in word:
                    score = score + letter_points_dict[user_input]
                    print("Correct guess!")
                else:
                    score = score - letter_points_dict[user_input]
                    print("Wrong guess!")

                guess_list.append(user_input)

            if is_game_over(word, guess_list):
                print('The word was ' + word)
                print('Game Over, your score is: ' + str(score))

                # TODO 9: Display a message based on score
                if score >= 50:
                    print("Excellent score!")
                elif score >= 20:
                    print("Good job!")
                else:
                    print("Keep practising!")

                break
        else:
            print("Invalid input. Please enter one letter.")


main()
```
